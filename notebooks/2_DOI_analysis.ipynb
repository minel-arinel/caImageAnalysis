## Tank temporal pack (single HDF5)

Writes **`temporal_pack.h5`** in the data folder with:
- Group **`metadata/`**: `fish`, `age`, `stimulus`, `concentration`, `plane`, `region`, `roi_index`, `pulse_frames`
- Datasets **`temporal`**, **`raw_temporal`**, **`norm_temporal`**, **`raw_norm_temporal`**: each `(n_rois, T)` float32

Run the next cell once (after `tank` exists). Use `overwrite=True` to recompute each fish's traces. `sort_fish_by_id=True` processes fish in numeric order.
